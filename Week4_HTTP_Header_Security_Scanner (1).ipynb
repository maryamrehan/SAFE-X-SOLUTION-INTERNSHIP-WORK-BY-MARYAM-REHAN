```python
import requests

# Read URLs from urls.txt
with open("urls.txt", "r") as f:
    urls = [line.strip() for line in f if line.strip()]

security_headers = {
    "Content-Security-Policy": "CSP",
    "Strict-Transport-Security": "HSTS",
    "X-Content-Type-Options": "X-Content-Type-Options",
    "X-Frame-Options": "X-Frame-Options",
    "Referrer-Policy": "Referrer-Policy",
    "Permissions-Policy": "Permissions-Policy"
}

for url in urls:
    print("=" * 60)
    print(f"URL: {url}")
    print("=" * 60)

    try:
        response = requests.get(url, timeout=10)

        print(f"Status Code: {response.status_code}\n")
        print("Security Header Analysis\n")

        for header, name in security_headers.items():

            # HSTS is only applicable to HTTPS
            if header == "Strict-Transport-Security" and not url.startswith("https://"):
                print("[NOT EVALUATED] HSTS (Target uses HTTP)")

            elif header in response.headers:
                print(f"[PRESENT] {name}: {response.headers[header]}")

            else:
                print(f"[MISSING] {name}")

    except requests.RequestException as e:
        print(f"[ERROR] Could not connect to {url}: {e}")

    print()
```

```python
import requests
from datetime import datetime

SECURITY_HEADERS = {
    "Content-Security-Policy": {
        "severity": "High",
        "remediation": "Configure a suitable Content-Security-Policy."
    },
    "Strict-Transport-Security": {
        "severity": "Medium",
        "remediation": "Enable HSTS when the application is served over HTTPS."
    },
    "X-Content-Type-Options": {
        "severity": "Medium",
        "remediation": "Set X-Content-Type-Options to nosniff."
    },
    "X-Frame-Options": {
        "severity": "Medium",
        "remediation": "Set X-Frame-Options to SAMEORIGIN or DENY."
    },
    "Referrer-Policy": {
        "severity": "Low",
        "remediation": "Configure an appropriate Referrer-Policy."
    },
    "Permissions-Policy": {
        "severity": "Low",
        "remediation": "Configure Permissions-Policy according to application requirements."
    }
}


def scan_url(url):
    results = []

    try:
        response = requests.get(url, timeout=10, allow_redirects=True)

        headers = {k.lower(): v for k, v in response.headers.items()}

        for header, info in SECURITY_HEADERS.items():
            value = headers.get(header.lower())

            if value:
                results.append({
                    "URL": url,
                    "Endpoint": response.url,
                    "Header": header,
                    "Vulnerable": "No",
                    "Severity": "None",
                    "Evidence": value,
                    "Remediation": "Header is present."
                })
            else:
                # HSTS is not treated as a missing-header finding on HTTP
                if header == "Strict-Transport-Security" and response.url.startswith("http://"):
                    results.append({
                        "URL": url,
                        "Endpoint": response.url,
                        "Header": header,
                        "Vulnerable": "No",
                        "Severity": "Informational",
                        "Evidence": "HSTS not evaluated because target uses HTTP.",
                        "Remediation": info["remediation"]
                    })
                else:
                    results.append({
                        "URL": url,
                        "Endpoint": response.url,
                        "Header": header,
                        "Vulnerable": "Yes",
                        "Severity": info["severity"],
                        "Evidence": "Header is missing.",
                        "Remediation": info["remediation"]
                    })

        return response.status_code, results

    except Exception as e:
        return None, [{
            "URL": url,
            "Endpoint": url,
            "Header": "Connection",
            "Vulnerable": "Unknown",
            "Severity": "N/A",
            "Evidence": str(e),
            "Remediation": "Verify that the authorized lab target is running."
        }]


# Multiple authorized lab URLs
urls = [
    "http://127.0.0.1:3000"
]

all_results = []

for url in urls:
    status, results = scan_url(url)
    all_results.extend(results)

print("=" * 70)
print("SECURITY HEADER SCANNER")
print("Scan Time:", datetime.now())
print("=" * 70)

for item in all_results:
    print(f"\nURL: {item['URL']}")
    print(f"Endpoint: {item['Endpoint']}")
    print(f"Header: {item['Header']}")
    print(f"Vulnerable: {item['Vulnerable']}")
    print(f"Severity: {item['Severity']}")
    print(f"Evidence: {item['Evidence']}")
    print(f"Remediation: {item['Remediation']}")
    print("-" * 70)
```

```python
def confirm_findings(url):
    status, results = scan_url(url)

    confirmed = []

    for item in results:
        # A finding is confirmed only when:
        # 1. The request succeeded
        # 2. The header is actually absent
        # 3. The finding is not HSTS on plain HTTP

        if (
            item["Vulnerable"] == "Yes"
            and item["Evidence"] == "Header is missing."
            and not (
                item["Header"] == "Strict-Transport-Security"
                and url.startswith("http://")
            )
        ):
            item["Confirmation"] = "Confirmed missing header"
        else:
            item["Confirmation"] = "No vulnerability confirmed"

        confirmed.append(item)

    return status, confirmed


status, confirmed_results = confirm_findings(
    "http://127.0.0.1:3000"
)

print("=" * 70)
print("REFINED / CONFIRMED SECURITY HEADER SCAN")
print("=" * 70)
print("HTTP Status:", status)

for item in confirmed_results:
    print(f"\nHeader: {item['Header']}")
    print(f"Vulnerable: {item['Vulnerable']}")
    print(f"Severity: {item['Severity']}")
    print(f"Evidence: {item['Evidence']}")
    print(f"Confirmation: {item['Confirmation']}")
```
